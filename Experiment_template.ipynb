```python
import pandas as pd
import shutil
```

```python
def cp_files_to(lst_paths_from=[], path_to='', prefix=''):
    """
        Copy the files from the list of paths 
        to a path direction
    """
    for f in lst_paths_from:
        shutil.copy(f"{prefix}{f}", path_to)
```

```python
# Analysis file with paths
df_own_analysis = pd.read_csv('Analysis/log_errors.csv')
# Get own hallucinations lst_paths
lst_own_hallucinations = df_own_analysis[df_own_analysis["message"].str.contains('has no') | df_own_analysis["message"].str.contains('unexpected')]["old"].values
# Get own borderline lst_paths
lst_own_borderlines = df_own_analysis[df_own_analysis["message"].str.contains('must') | df_own_analysis["message"].str.contains('type')]["old"].values
```

```python
# Move files to path
h_p = 'Data/Own/Hallucinations/'
b_p = 'Data/Own/Borderline/'
prefix = '../../RP4 2023/LLM-Creativity/'
cp_files_to(lst_own_hallucinations, h_p, prefix)
cp_files_to(lst_own_borderlines, b_p, prefix)
```

```python
def template(exp_id=0, code_document='', dir_prefix=''):
    """
        Writes the buggy code in the style of a prompt
        template experiment to disk
    """
    code_document_idx = code_document.find("document")
    file_to_write = f"{dir_prefix}{code_document[code_document_idx:]}_exp_{exp_id}"
    if exp_id == 0:
        with open(file_to_write, 'w') as f:
            f.write()
    return 
```
